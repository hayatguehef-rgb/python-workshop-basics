```python
import pandas as pd

data = {
    'Product': ['A', 'B', 'C', 'A', 'B', 'C', 'A', 'B'],
    'Region': ['North', 'South', 'East', 'West', 'North', 'South', 'East', 'West'],
    'Sales': [250, 400, 150, 300, 450, 200, 350, 500]
}

df = pd.DataFrame(data)

print("--- جدول المبيعات الأصلي ---")
print(df)
print("\n" + "="*30 + "\n")

total_sales_by_product = df.groupby('Product')['Sales'].sum()
print("1. إجمالي المبيعات لكل منتج:")
print(total_sales_by_product)
print("\n" + "="*30 + "\n")

avg_sales_by_region = df.groupby('Region')['Sales'].mean()
print("2. متوسط المبيعات لكل منطقة:")
print(avg_sales_by_region)
print("\n" + "="*30 + "\n")

product_stats = df.groupby('Product')['Sales'].agg(['mean', 'max'])
print("3. المتوسط والحد الأقصى للمبيعات لكل منتج:")
print(product_stats)
```
