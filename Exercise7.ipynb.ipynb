```python
import pandas as pd

fruits_prices = pd.Series({
    'Apple': 3.0,
    'Banana': 1.5,
    'Orange': 2.5,
    'Mango': 5.0,
    'Grapes': 4.0
})

print("1. Original Fruit Prices Series:")
print(fruits_prices)
print("-" * 30)

third_fruit_price = fruits_prices.iloc[2]
print("2. Price of the 3rd fruit (Orange):", third_fruit_price)
print("-" * 30)

increased_prices = fruits_prices * 1.1
print("3. Prices after 10% increase:")
print(increased_prices)
print("-" * 30)

avg_price = increased_prices.mean()
most_expensive_fruit = increased_prices.idxmax()

print(f"4. Average Price: {avg_price:.2f}")
print(f"   Most Expensive Fruit: {most_expensive_fruit} (${increased_prices[most_expensive_fruit]:.2f})")
```
