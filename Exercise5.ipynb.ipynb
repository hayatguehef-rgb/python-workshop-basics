```python

with open("mymodule.py", "w") as f:
    f.write('''def create_dictionary(keys, values):
    \"\"\"
    Takes two lists (keys and values) and returns a dictionary.
    Returns an error message if lists have different lengths.
    \"\"\"
    if len(keys) != len(values):
        return "Error: Lists must have the same length."

    return dict(zip(keys, values))
''')


import mymodule


keys_list = ["name", "age", "city"]
values_list = ["Alice", 25, "New York"]

result_dict = mymodule.create_dictionary(keys_list, values_list)


print("Generated Dictionary:", result_dict)
```
