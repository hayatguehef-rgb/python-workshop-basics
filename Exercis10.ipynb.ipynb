```python
import pandas as pd
import matplotlib.pyplot as plt

revenue_data = {
    'Quarter': ['Q1', 'Q2', 'Q3', 'Q4'],
    'Revenue': [12000, 15000, 13500, 18000]
}
df_revenue = pd.DataFrame(revenue_data)

print("--- جدول الإيرادات الفصلية ---")
print(df_revenue)
print("\n" + "="*40 + "\n")


plt.figure(figsize=(8, 4))
plt.plot(df_revenue['Quarter'], df_revenue['Revenue'], marker='o', color='blue', linestyle='-', linewidth=2)
plt.title('Quarterly Revenue Trend')
plt.xlabel('Quarter')
plt.ylabel('Revenue ($)')
plt.grid(True)
plt.show()


plt.figure(figsize=(8, 4))
plt.bar(df_revenue['Quarter'], df_revenue['Revenue'], color='green', alpha=0.7)
plt.title('Quarterly Revenue Comparison')
plt.xlabel('Quarter')
plt.ylabel('Revenue ($)')
plt.show()

scores_data = {
    'Student': ['Alice', 'Bob', 'Charlie', 'David', 'Eve', 'Frank', 'Grace', 'Hannah', 'Ian', 'Jack'],
    'Score': [78, 85, 92, 65, 88, 79, 95, 90, 82, 74]
}
df_scores = pd.DataFrame(scores_data)

print("--- جدول درجات الطلاب ---")
print(df_scores)
print("\n" + "="*40 + "\n")

plt.figure(figsize=(8, 4))
plt.hist(df_scores['Score'], bins=5, color='orange', edgecolor='black', alpha=0.8)
plt.title('Distribution of Student Test Scores')
plt.xlabel('Score Ranges')
plt.ylabel('Frequency (Number of Students)')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```
