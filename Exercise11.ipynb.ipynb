```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = sns.load_dataset('iris')

print("--- أول 5 صفوف من بيانات Iris ---")
print(df.head())
print("\n" + "="*50 + "\n")

virginica_df = df[df['species'] == 'virginica']
print("1. أول 5 صفوف لنوع 'virginica':")
print(virginica_df.head(5))
print("\n" + "="*50 + "\n")

min_petal_width = df['petal_width'].min()
max_petal_width = df['petal_width'].max()
print(f"2. أقل عرض للبتلة: {min_petal_width}")
print(f"   أقصى عرض للبتلة: {max_petal_width}")
print("\n" + "="*50 + "\n")


avg_sepal_length = df.groupby('species')['sepal_length'].mean()
highest_species = avg_sepal_length.idxmax()
print("3. متوسط طول السبلة لكل نوع:")
print(avg_sepal_length)
print(f"\nالنوع الذي يمتلك أعلى متوسط لطول السبلة هو: **{highest_species}**")
print("\n" + "="*50 + "\n")

plt.figure(figsize=(8, 5))
sns.scatterplot(data=df, x='petal_length', y='petal_width', hue='species', palette='deep')
plt.title('Relationship between Petal Length and Petal Width')
plt.xlabel('Petal Length')
plt.ylabel('Petal Width')
plt.grid(True, alpha=0.5)
plt.show()

avg_petal_length = df.groupby('species')['petal_length'].mean()

plt.figure(figsize=(8, 5))
avg_petal_length.plot(kind='bar', color=['skyblue', 'lightgreen', 'salmon'], edgecolor='black')
plt.title('Average Petal Length by Species')
plt.xlabel('Species')
plt.ylabel('Average Petal Length')
plt.xticks(rotation=0)
plt.grid(axis='y', alpha=0.5)
plt.show()
```
