```python
import numpy as np

arr = np.arange(10, 51, 5)
print("1. Array:", arr)

sliced_arr = arr[2:6]
print("2. Sliced array (index 2 to 5):", sliced_arr)

mean_val = np.mean(arr)
std_val = np.std(arr)
print(f"3. Mean: {mean_val}, Standard Deviation: {std_val}")

matrix = np.arange(1, 7).reshape(2, 3)
print("4. 2x3 Matrix:\n", matrix)

element = matrix[1, 2]
print("Element at row 1, column 2:", element)
```
