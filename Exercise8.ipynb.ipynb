```python
import pandas as pd

data = {
    'Math': [85, 72, 90, 65, 88],
    'English': [78, 88, 92, 70, 86],
    'Science': [92, 80, 85, 75, 95]
}
students = ['Alice', 'Bob', 'Charlie', 'David', 'Eve']

df = pd.DataFrame(data, index=students)

print("--- DataFrame الكامل ---")
print(df)
print("\n" + "="*30 + "\n")

print("1. أول 3 صفوف (head):")
print(df.head(3))
print("\n" + "="*30 + "\n")

print("2. درجات الرياضيات فقط:")
print(df['Math'])
print("\n" + "="*30 + "\n")

print("3. الطلاب الحاصلون على أكثر من 80 في الرياضيات:")
math_above_80 = df[df['Math'] > 80]
print(math_above_80)
print("\n" + "="*30 + "\n")

print("4. الطلاب الحاصلون على أكثر من 85 في Math AND English:")
top_students = df[(df['Math'] > 85) & (df['English'] > 85)]
print(top_students)
```
